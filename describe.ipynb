```python
import pandas as pd

data = {
    "Patient_ID": [101,102,103,104,105,106,107,108,109,110],
    "Age": [25,45,35,60,50,28,65,40,55,30],
    "Gender": ["Male","Female","Male","Female","Male",
               "Female","Male","Female","Male","Female"],
    "Department": ["Cardiology","Neurology","Cardiology","Orthopedic",
                   "Neurology","Cardiology","Orthopedic","Neurology",
                   "Cardiology","Orthopedic"],
    "City": ["Nagpur","Pune","Nagpur","Mumbai","Pune",
             "Nagpur","Mumbai","Pune","Nagpur","Mumbai"],
    "Treatment_Cost": [50000,35000,60000,45000,30000,
                       55000,70000,40000,65000,50000],
    "Hospital_Stay": [5,3,6,4,2,5,7,3,6,4]
}

df = pd.DataFrame(data)

df
```

```python
# har department mien kitne patient hain

df.groupby("Department")
```

```python
df.groupby("Department")["Patient_ID"].size()
```

```python
df.groupby("Department")["Patient_ID"].count()
```

```python
#har departments me average treatment cost kkitni hai
df.groupby("Department")["Treatment_Cost"].mean()
```

```python
df.groupby("Department")["Treatment_Cost"].sum()
```

1. Maximum Treatment Cost in each Department

```python
df.groupby("Department")["Treatment_Cost"].max()
```

2. Patient_ID having Maximum Treatment Cost

```python
df.loc[df.groupby("Department")["Treatment_Cost"].idxmax(),["Department", "Patient_ID", "Treatment_Cost"]]
```

```python
df.groupby("Department").agg({
    "Patient_ID": "count",
    "Age": "mean",
    "Treatment_Cost": "sum",
    "Hospital_Stay": "mean"
})
```

```python
# 3. Minimum Treatment Cost
df.groupby("Department")["Treatment_Cost"].min()
```

```python
# Patient_ID having Minimum Treatment Cost
df.loc[df.groupby("Department")["Treatment_Cost"].idxmin(),
       ["Department", "Patient_ID", "Treatment_Cost"]]
```

Maximum value → max()
Maximum value's row → idxmax()

Minimum value → min()
Minimum value's row → idxmin()

```python
# Oldest Patient in Each Department
df.loc[df.groupby("Department")["Age"].idxmax(),
       ["Department", "Patient_ID", "Age"]]
```

```python
# 14. Minimum Hospital Stay Patient
df.loc[df.groupby("Department")["Hospital_Stay"].min(),
       ["Department", "Patient_ID", "Hospital_Stay"]]
```

```python
report = df.groupby("Department").agg(
    Total_Patients=("Patient_ID", "count"),
    Average_Treatment_Cost=("Treatment_Cost", "mean"),
    Maximum_Treatment_Cost=("Treatment_Cost", "max")
)

print(report)
```

```python
report = df.groupby("Department").agg(
    Patient_Count=("Patient_ID", "count"),
    Average_Age=("Age", "mean"),
    Total_Treatment_Cost=("Treatment_Cost", "sum"),
    Maximum_Treatment_Cost=("Treatment_Cost", "max"),
    Average_Hospital_Stay=("Hospital_Stay", "mean")
)

print(report)
```
